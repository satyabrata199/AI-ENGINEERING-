````python

def maxnum(num):
    m = num[0]
    for e in num:
        if e>m:
            m = e
    return m
    

numbers = [10, 4, 25, 7, 18, 2]
print(maxnum(numbers))
````

````python

def twosum(num , tg):
    l = 0
    r = len(num)-1
    while(l<r):
        con = num[l]+num[r]
        if con == tg:
            return [num[l],num[r]]
        if con < tg:
            l+=1
        if con > tg:
            r-=1

    return 'None'
            


numbers = [2, 7, 11, 15]
target = 9

print(twosum(numbers,target))
````

````python
from collections import Counter
def check(s1,s2):
    if len(s1) != len(s2):
        return False 

    if Counter(s1) == Counter(s2):
        return True

    else:
        return False
        

s1 = 'hlo'
s2 = 'lch'
check(s1,s2)
````

````python
"""Python's id() function returns an identity value that is unique among objects whose lifetimes overlap.
In CPython, this is commonly the object's memory address, 
but you should not assume that for every Python implementation."""

a = [1,2,3]
b = a
c = [1,2,3]

print(id(a))
print(id(b))
print(id(c))

print(f"{a is b}, {b is c}, {c is a}")
print(f"{a == b} , {b == c} , {c == a}")
print(f"{id(a) == id(c)},{id(b) == id(c)}")
````
