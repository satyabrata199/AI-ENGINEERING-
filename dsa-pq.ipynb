```python
#Problem 6 — Left Rotate Array by k Places
class numrotate:
    def rotation(self,a,left,right):
        while(left<right):
            a[left],a[right] = a[right],a[left]
            left+=1
            right-=1

        return a
        
obj = numrotate()
a = [1,2,3,4,5,6,7,8]
k = 2
r = len(a)-1
obj.rotation(a,0,k-1)
obj.rotation(a,k,r)
print(obj.rotation(a,0,r))
```

```python
#Problem 11 Union of Two Sorted Arrays
def union(a,b):
    s = set({})
    for i,j in zip(a,b):
        s.add(i)
        s.add(j)
    return s
a = [1,2,3,4]
b = [3,4,5,6]
c = union(a,b)
print(c)
#in short 
s1=set(a)
s2 = set(b)
print(s1.union(s2))
```

```python
# Problem 12 Intersection of Two Sorted Arrays
#Short hand method 
a = {1,2,3,4,5}
b = {3,4,5,6,7}
print(a.intersection(b))
#long one 
def intersection(a,b):
    c=[]
    for e in a:
        if e in b:
            c.append(e)

    return c
a = [1,2,3,4,5]
b = [3,4,5,6,7]
c = intersection(a,b)
print(c)
```

```python
#valid palindrome

def num_pal(num):
    temp = num
    rev = 0
    while num > 0:
        rem = num % 10
        rev = (rev * 10) + rem
        num = num // 10

    if temp == rev:
        return 'valid palindrome'

    return 'not palindrome'

def str_pal(st):
    if st == st[::-1]:
        return 'valid palindrome'
    return 'not valid palendrome'

r = num_pal(11211)
r1 = str_pal('madam')
print(r,r1)
```
